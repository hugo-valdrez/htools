```python
dishes = [
    {"id": 1, "name": "Bolognese", "protein": "Beef"},
    {"id": 2, "name": "Carbonara", "protein": "Pork"},
    {"id": 3, "name": "Tofu Rice Bowl", "protein": "Tofu"},
    {"id": 4, "name": "Tofu Stir Fry", "protein": "Tofu"},
    {"id": 5, "name": "Baked Fish & Potatoes", "protein": "Fish"},
    {"id": 6, "name": "Lentil Stew", "protein": "Lentils"},
    {"id": 7, "name": "Pasta e Fagioli", "protein": "Beans"},
    {"id": 8, "name": "Beef Risotto", "protein": "Beef"},
    {"id": 9, "name": "Steak and Mash", "protein": "Beef"},
    {"id": 10, "name": "Tofu Chickpea Salad", "protein": "Tofu"},
    {"id": 11, "name": "Tofu Pasta", "protein": "Tofu"},
    {"id": 12, "name": "Fish & Rice", "protein": "Fish"}
]
```

```python
from abc import ABC, abstractmethod

class BaseCondition(ABC):
    @abstractmethod
    def satisfies_condition(self, **kwargs) -> bool:
        pass
```

```python
class ConditionsManager:
    def __init__(self, conditions: list[BaseCondition]):
        self.conditions = conditions

    def check_conditions(self, meal_selection):
        for condition in self.conditions:
            if not condition.satisfies_condition(meal_selection=meal_selection):
                return False
        return True
```

```python
class mandatory_MealPlanSize(BaseCondition):
    def __init__(self, meal_plan_size: int = 7):
        self.meal_plan_size = meal_plan_size

    def satisfies_condition(self, meal_selection, **kwargs):
        return len(meal_selection) == self.meal_plan_size
```

```python
class mandatory_TofuDishes(BaseCondition):
    def __init__(self, tofu_dishes_size: int = 1):
        self.tofu_dishes_size = tofu_dishes_size

    def satisfies_condition(self, meal_selection, **kwargs):
        return sum(1 for dish in meal_selection if dish["protein"] == "Tofu") == self.tofu_dishes_size
```

```python
import random

conditions_manager = ConditionsManager([
    mandatory_MealPlanSize(meal_plan_size=7),
    mandatory_TofuDishes(tofu_dishes_size=1),
])

def solve_meal_plan(current_plan, remaining_dishes):
    if conditions_manager.check_conditions(meal_selection=current_plan):
        return current_plan

    if len(current_plan) >= 7 or not remaining_dishes:
        return None

    indices = list(range(len(remaining_dishes)))
    random.shuffle(indices)

    for idx in indices:
        selection = remaining_dishes[idx]
        new_remaining = remaining_dishes[:idx] + remaining_dishes[idx+1:]

        result = solve_meal_plan(current_plan + [selection], new_remaining)
        if result is not None:
            return result

    return None

final_meal_plan = solve_meal_plan([], dishes)
```
